```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import ovrlpy
from pathlib import Path
import random
```

```python
random.seed(42)


coordinate_path = f"/dh-projects/ag-ishaque/raw_data/vizgen-merfish/vz-receptor-showcase/Slice2/Replicate1/detected_transcripts_S2R1.csv"
coordinate_df=pd.read_csv(coordinate_path, sep=",", usecols=[2, 3, 4, 8], header=0)[::1]


coordinate_df.columns = ["x", "y", "z", "gene"]
coordinate_df = coordinate_df[~coordinate_df.gene.str.contains("Blank")]

coordinate_df.x -= coordinate_df.x.min()
coordinate_df.y -= coordinate_df.y.min()
coordinate_df.z -= coordinate_df.z.min()


coordinate_df.gene = coordinate_df.gene.astype("str").astype("category")
```

```python
ovrlpy_50 = ovrlpy.Ovrlp(coordinate_df, n_workers=8, random_state=42,n_components=50)

ovrlpy_50.analyse()

fig_50 = ovrlpy.plot_signal_integrity(ovrlpy_50)


fig_50.axes[0].invert_xaxis()
fig_50.axes[0].invert_yaxis()
```

```python
fig_50.savefig('MERFISH_PC50_rotated.svg', format='svg', dpi=300)
```

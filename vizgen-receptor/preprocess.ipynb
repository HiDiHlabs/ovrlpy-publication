```python
from pathlib import Path

coordinate_path = Path(
    "/dh-projects/ag-ishaque/raw_data/vizgen-merfish/vz-receptor-showcase/Slice2/Replicate1"
)

out = Path(".")
```

We will need the assignment from transcript to cell which can be achieved by `vpt partition-transcripts` but first we need to prepare the segmentation masks to a compatible format.

Following commands must be executed in a shell with `vpt` installed.

```python
(
    "vpt --processes 4 convert-geometry "
    f"--input-boundaries {coordinate_path / 'cell_boundaries' / 'feature_data_'} "
    f"--output-boundaries {out / 'cell_boundaries.parquet'} "
    f"--id-mapping-file {out / 'id_mapping.csv'} "
)
```

```python
(
    "vpt --processes 4 partition-transcripts "
    f"--input-boundaries {out / 'cell_boundaries.parquet'} "
    f"--input-transcripts {coordinate_path / 'detected_transcripts_S2R1.csv'} "
    f"--output-entity-by-gene {out / 'cell_by_gene.csv'} "
    f"--output-transcripts {out / 'detected_transcripts.parquet'} "
)
```

```python
(
    "vpt --processes 4 derive-entity-metadata "
    f"--input-boundaries {out / 'cell_boundaries.parquet'} "
    f"--output-metadata {out / 'cell_metadata.csv'} "
)
```
